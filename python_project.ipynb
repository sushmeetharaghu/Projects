# **TASK 2**

```python
# importing pandas library
import pandas as pd

# reading already saved csv files
t1=pd.read_csv('project_dataframe.csv')

# to calculate running average, initiating r_sum and r_count equal to zero
r_sum=0
r_count=0

#using for loop to iterate from indexes 0-13 of dataframe
for i in range(len(t1)):
    #assigning cost value to intial proj_cost
  proj_cost=t1.loc[i,'Cost']
    #checking if assigned proj_cost from cost is NA
  if pd.isna(proj_cost):
    # if NA, then r_avg is calculated as r_sum/r_count
    r_avg=r_sum/r_count
      #assigning the running average to cost column
    t1.loc[i,'Cost']=r_avg
  else:
      #if the value is not NA then it adds to the project cost value and count is incremented by 1
    r_sum += proj_cost
    r_count=r_count+1

#setting cost column as int type 
t1['Cost']=t1.Cost.astype(int)

#printing the dataframe
print(t1)

# t1 dataframe to a .csv file eliminating index on csv file
t1.to_csv('t1.csv',index=False)
```

# **TASK 3**

```python
# importing pandas and numpy
import pandas as pd

# reading already saves employee_dataframe.csv file
t2=pd.read_csv('employee_dataframe.csv')

#defining two column names 'First name','Last name' and spliting name column based on ' ' ,
# with no. of splits n=1 and expand=True to return as two columns
t2[['First name','Last name']]=t2['Name'].str.split(' ',n=1,expand=True)

#popping the first name column
first_name_column=t2.pop('First name')

#inserting the popped column in 1st integer position of columns
t2.insert(1,'First name',first_name_column)

#popping the last name column
last_name_column=t2.pop('Last name')

#inserting the popped column in 2st integer position of columns
t2.insert(2,'Last name',last_name_column)

#dropping name column from dataframe
t2 = t2.drop(columns=['Name'])

#printing the dataframe after droping name column
print(t2)

# t2 dataframe to a .csv file eliminating index on csv file
t2.to_csv('t2.csv',index=False)
```

# **TASK 4**

```python
# importing pandas and numpy
import pandas as pd

# reading already saved csv files
t2=pd.read_csv('t2.csv')
t1=pd.read_csv('t1.csv')
t3=pd.read_csv('seniority_level_dataframe.csv')

#merging all the three dataframes using merge function performing inner joins
t4=pd.merge(t1,t2,on='ID',how='inner')
final=pd.merge(t4,t3,on='ID',how='inner')

#printing the final merged dataframe
print(final)

# final dataframe to a .csv file eliminating index on csv file
final.to_csv('final.csv',index=False)
```

# **TASK 5**

```python
# importing pandas and numpy
import pandas as pd
import numpy as np

# reading already saved csv file
final=pd.read_csv('final.csv')

#using a where function, which checks the status column is equal to 'Fisnished', if so calculates bonus, if not equal returns zero
final['bonus']=np.where(final['Status']=='Finished',final['Cost']*0.05,0).astype(int)

#printing the final dataframe
print(final)

# final dataframe to a .csv file eliminating index on csv file
final.to_csv('final.csv',index=False)

```

# **TASK 6**

```python
# importing pandas and numpy
import pandas as pd
import numpy as np

# reading already saved csv file
final=pd.read_csv('final.csv')

#using a where function, which checks the status column is equal to 'Failed', if so depromotes 1 designation level, else returns the same level
final['Designation level']=np.where(final['Status']=='Failed',final['Designation level']-1,final['Designation level'])
final = final[final['Designation level'] <= 4]

#printing the final dataframe
print(final)

# final dataframe to a .csv file eliminating index on csv file
final.to_csv('final.csv',index=False)
```

```python
# importing pandas and numpy
import pandas as pd
import numpy as np

# reading already saved csv file
final=pd.read_csv('final.csv')

# using groupby function using ID & First name columns in final dataframe and performing sum aggregation of cost column
totalprojcost=final.groupby(['ID','First name']).agg(**{'total cost':('Cost','sum')}).reset_index()

#printing the dataframe
print(totalprojcost)

# totalpro dataframe to a .csv file eliminating index on csv file
totalprojcost.to_csv('totalprojcost.csv',index=False)
```
